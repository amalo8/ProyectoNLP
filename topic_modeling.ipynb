### TOPIC MODELING HOTELES EN ESPAÑOL

```python
import pandas as pd
import os

# 1. Cargar el dataset limpio
df_limpio = pd.read_csv('datos/procesados/df_comentarios_limpio.csv')

# 2. Filtrar solo los comentarios de clientes de España
df_espana = df_limpio[df_limpio['nacionalidad'] == 'españa']

print(f"Total de comentarios de España: {len(df_espana)}")

# 3. Crear una carpeta para guardar los mini datasets (opcional pero recomendado para mantener el orden)
carpeta_destino = 'mini_datasets_hoteles'
if not os.path.exists(carpeta_destino):
    os.makedirs(carpeta_destino)

# NOTA: Asegúrate de que el nombre de la columna que contiene el hotel es correcto. 
columna_hotel = 'nombre_del_hotel' 

# 4. Agrupar por hotel y guardar un CSV para cada uno
for nombre_hotel, df_hotel in df_espana.groupby(columna_hotel):
    
    # Limpiamos el nombre del hotel para que sea un nombre de archivo válido 
    # (reemplazamos espacios por guiones bajos y quitamos caracteres conflictivos si los hubiera)
    nombre_archivo_seguro = str(nombre_hotel).replace(' ', '_').replace('/', '-')
    ruta_archivo = f"{carpeta_destino}/comentarios_{nombre_archivo_seguro}.csv"
    
    # Guardamos el mini dataset
    df_hotel.to_csv(ruta_archivo, index=False)
    
print(f"¡Listo! Se han guardado {df_espana[columna_hotel].nunique()} mini datasets en la carpeta '{carpeta_destino}'.")
```
